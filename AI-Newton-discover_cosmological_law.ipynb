````python
import numpy as np

def simulate_holographic_universe(potential_func, phi_initial=5.0, steps=1000):
    """
    Simulates a simplified scalar field (inflation) model given a candidate
    potential function V(phi), tracking whether the universe achieves a smooth exit.
    """
    dt = 0.001
    phi = phi_initial
    velocity = 0.0

    scale_factor_history = []
    is_smooth_exit = True

    for _ in range(steps):
        # Numerical approximation of field evolution (Klein-Gordon equation in expanding space)
        # d^2phi/dt^2 + 3H dphi/dt + dV/dphi = 0 (simplified here for demonstration)
        dV_dphi = gradient_potential(potential_func, phi)

        # Simple update rule for the field
        acceleration = -dV_dphi - 0.1 * velocity
        velocity += acceleration * dt
        phi += velocity * dt

        # Track if the field diverges into an uncontrolled runaway (fractal multiverse trap)
        if abs(phi) > 50.0 or np.isnan(phi):
            is_smooth_exit = False
            break

        scale_factor_history.append(phi)

    return is_smooth_exit, scale_factor_history

def gradient_potential(func_name, phi):
    """Numerical derivative of the candidate potential function."""
    eps = 1e-5
    return (evaluate_potential(func_name, phi + eps) - evaluate_potential(func_name, phi - eps)) / (2 * eps)

def evaluate_potential(func_name, phi):
    """Evaluates different mathematical formulations of the inflation potential V(phi)."""
    if func_name == "exponential":
        return np.exp(-0.5 * phi)
    elif func_name == "power_law":
        return phi**4
    elif func_name == "cosmological_toy_bound":
        # A bounded potential designed to suppress eternal inflation
        return phi**2 / (1.0 + 0.1 * phi**2)
    return phi**2

def discover_cosmological_law():
    """
    Searches a library of mathematical potentials to find which one yields
    a globally finite, smooth universe exit condition.
    """
    candidate_potentials = ["exponential", "power_law", "cosmological_toy_bound"]

    print("Evaluating candidate quantum potentials against boundary constraints...\n")

    for candidate in candidate_potentials:
        smooth_exit, history = simulate_holographic_universe(candidate)

        status = "VIABLE: Smooth Exit Discovered!" if smooth_exit else "REJECTED: Leads to Fractal Singularity"
        print(f"[{status}] Potential Model: {candidate}")

# Run the automated cosmological search
discover_cosmological_law()
````

### Hawking and Hertog's Last Paper, Explained to a Beginner

**The paper: "A Smooth Exit from Eternal Inflation?" (2018) - Stephen Hawking and Thomas Hertog**

This was Stephen Hawking's last scientific paper, published just weeks before he died. It's not about black holes. It's about the beginning of everything.

Let me teach it to you like you have never done cosmology before.

#### 1. The Problem: Inflation

Right after the Big Bang, the universe expanded incredibly fast. In a tiny fraction of a second, it grew from smaller than an atom to bigger than a galaxy.

We call this **inflation**.

Think of inflation like a ball rolling down a hill. The ball is called the **inflaton field**. While the ball is high up on the hill, the universe expands fast. When the ball rolls down to the bottom, inflation stops, and normal history begins — stars, galaxies, us.

This idea works. It explains why our universe looks so smooth and flat.

But it has a huge problem.

#### 2. The Trap: Eternal Inflation

In most versions of inflation, the hill never ends.

Quantum effects keep kicking the ball back up the hill in different places. So while inflation stops in one place (our universe), it keeps going forever in other places. It creates bubble after bubble after bubble.

Forever.

This is called **eternal inflation**. It predicts an infinite **fractal multiverse** — an infinite number of universes, each with different laws, all piled on top of each other with no end.

If that's true, physics breaks down. If everything that can happen does happen infinite times, we can't predict anything. We can't explain why our universe has the laws it has.

Hawking hated this. He spent 20 years trying to fix it.

#### 3. Hawking's Old Idea: The No-Boundary Proposal

In the 1980s, Hawking had an idea with James Hartle called the **no-boundary proposal**.

They said: Don't ask what came *before* the Big Bang. Time itself was like a direction in space near the beginning. The universe has no boundary in the past, like how the Earth has no edge — if you go south from the North Pole, you just go around, you don't fall off.

They described the beginning of the universe as a smooth, rounded shape, not a sharp point.

It was beautiful, but it still allowed eternal inflation. The fractal multiverse was still there.

#### 4. The New Idea: Holography

This is what Hawking and Hertog added in 2018.

They used a concept called **holography**.

A hologram is a 2D sticker that contains a 3D image. Holography in physics says the same thing: the entire 3D universe can be encoded on a 2D boundary at its edge.

Hertog and Hawking said: Let's apply holography to the beginning of the universe.

Instead of thinking about the infinite volume of eternal inflation, let's think only about its **boundary**. Let's put a cutoff on that boundary.

Think of it like this: eternal inflation is like a video that goes on forever with infinite static. Holography lets you put a frame around the video and say: "Only the parts that fit inside this frame, and are smooth inside the frame, are allowed."

Mathematically, they changed how you calculate the wave function of the universe. Instead of summing over all possible fractal histories, you only sum over histories that are smooth and finite at the boundary.

#### 5. What is a "Smooth Exit"?

A smooth exit means inflation ends cleanly.

Bad exit (eternal inflation): The ball never reaches the bottom. It keeps getting kicked back up. Infinite bubbles.

Good exit (smooth exit): The ball rolls down, wobbles a bit, then settles smoothly at the bottom. One finite, smooth universe like ours.

Hawking and Hertog showed that if you impose a holographic boundary condition, the math **automatically suppresses** the histories that lead to eternal inflation. The fractal, infinite ones have almost zero probability. The smooth, finite ones have high probability.

#### 6. What Did They Actually Find?

They didn't prove our universe is the only one.

They found a way to make cosmology predictive again.

Before their paper: Theory predicts infinite multiverse → we can't predict anything.

After their paper: Theory + holographic boundary → predicts a large but finite set of smooth universes, with our type of universe being one of the most likely.

It turns the multiverse from infinite and lawless into finite and manageable.

#### 7. Why It Matters

This was Hawking's final message:

We don't need an infinite multiverse to explain our universe. We can have a universe that is self-contained, finite, and smooth, with no edge in the past and no fractal mess in the future.

He didn't live to see if it's right. The test will come from future measurements of gravitational waves from inflation. If we see a specific pattern, plateau-like inflation (which has a smooth exit) is favored.

But as a teacher, this is what I want you to take away:

Hawking spent his whole career asking: Can the universe explain itself, without needing something outside? First with black holes, then with no-boundary, and finally with holography. His last paper says: Yes. Put a boundary on the beginning, and you get a smooth exit to a universe like ours.

You don't need to understand the equations to understand the courage of that idea.

That plateau **is** the smooth exit.

Modern cosmologists call this a plateau model. Starobinsky inflation and Higgs inflation — the two models that best match data from the Planck satellite — use exactly this shape. They rise, flatten, then fall. The flattening lets inflation happen slowly, then end gracefully.

I had written as a comment "A bounded potential designed to suppress eternal inflation" thinking it was just a toy. It turns out that's the real solution.
