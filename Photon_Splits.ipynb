```python
import numpy as np

# Define single-photon polarization basis states (Jones vectors)
H = np.array([1, 0], dtype=complex)  # Horizontal polarization
V = np.array([0, 1], dtype=complex)  # Vertical polarization

print("="*60)
print("1. TYPE-I SPDC: PARALLEL POLARIZATIONS")
print("="*60)

# In Type-I SPDC, daughter photons share parallel polarizations.
# Product state example: Both photons are horizontally polarized |H> ⊗ |H>
state_type_I_product = np.kron(H, H)
print("Type-I Product State (|H, H>):")
print(state_type_I_product)

# Superposition/Entangled Type-I state (e.g., using a pump polarized at 45°):
# |Psi_I> = (|H, H> + |V, V>) / sqrt(2)
state_type_I_entangled = (np.kron(H, H) + np.kron(V, V)) / np.sqrt(2)
print("\nType-I Entangled State ((|H,H> + |V,V>) / sqrt(2)):")
print(state_type_I_entangled)


print("\n" + "="*60)
print("2. TYPE-II SPDC: ORTHOGONAL POLARIZATIONS")
print("="*60)

# In Type-II SPDC, daughter photons have orthogonal polarizations[cite: 1].
# The standard polarization-entangled Bell state produced is:
# |Psi^+> = (|H, V> + |V, H>) / sqrt(2)[cite: 5]
state_type_II = (np.kron(H, V) + np.kron(V, H)) / np.sqrt(2)
print("Type-II Entangled State (Psi^+ = (|H,V> + |V,H>) / sqrt(2)):")
print(state_type_II)


print("\n" + "="*60)
print("3. COINCIDENCE MEASUREMENT SIMULATION")
print("="*60)

def polarizer_matrix(theta):
    """Returns the 2x2 Jones matrix for a linear polarizer oriented at angle theta."""
    cos_t = np.cos(theta)
    sin_t = np.sin(theta)
    # Projection operator for transmission axis |theta><theta|
    return np.array([
        [cos_t**2, cos_t*sin_t],
        [cos_t*sin_t, sin_t**2]
    ], dtype=complex)

# Test coincidence probability for the Type-II entangled state
# as a function of polarizer angles theta_1 and theta_2
theta_1 = np.pi / 4  # 45 degrees
theta_2 = np.pi / 4  # 45 degrees

P1 = polarizer_matrix(theta_1)
P2 = polarizer_matrix(theta_2)

# Combined two-photon measurement operator (tensor product)
P_total = np.kron(P1, P2)

# Calculate coincidence probability: P = <Psi | P_total | Psi>
coincidence_prob_II = np.real(np.vdot(state_type_II, P_total @ state_type_II))
print(f"Type-II Coincidence Probability at polarizers (45°, 45°): {coincidence_prob_II:.4f}")

# Compare with Type-I entangled state at the same angles
P_total_I = np.kron(polarizer_matrix(theta_1), polarizer_matrix(theta_2))
coincidence_prob_I = np.real(np.vdot(state_type_I_entangled, P_total_I @ state_type_I_entangled))
print(f"Type-I Coincidence Probability at polarizers (45°, 45°): {coincidence_prob_I:.4f}")
```
