## Lesson 2: Variables and basic types

A variable stores a value. Its type tells C++ what kind of value it holds: `int` for whole numbers, `double` for decimal numbers, `char` for one character, and `bool` for true-or-false values. `std::string` stores text.

```cpp
%%writefile lesson02.cpp
#include <iostream>
#include <string>

int main() {
    int apples = 4;
    double price = 1.25;
    char size = 'M';
    bool isRipe = true;
    std::string fruit = "apple";

    std::cout << "Fruit: " << fruit << '\n';
    std::cout << "Apples: " << apples << '\n';
    std::cout << "Price: $" << price << '\n';
    std::cout << "Size code: " << size << '\n';
    std::cout << "Ripe? " << std::boolalpha << isRipe << '\n';

    return 0;
}
```

Compile and run:

```python
!g++ -std=c++17 -Wall -Wextra lesson02.cpp -o lesson02 && ./lesson02
```

## Lesson 3: Getting input and doing arithmetic

`std::cin` reads a value typed by the user. The operators `+`, `-`, `*`, and `/` perform arithmetic. Here, the numbers are `double`, so division can produce a decimal result.

When the program prompts you, enter a number and press Enter; repeat for the second prompt.

```cpp
%%writefile lesson03.cpp
#include <iostream>

int main() {
    double firstNumber;
    double secondNumber;

    std::cout << "Enter a number: ";
    std::cin >> firstNumber;

    std::cout << "Enter another number: ";
    std::cin >> secondNumber;

    std::cout << "Sum: " << firstNumber + secondNumber << '\n';
    std::cout << "Difference: " << firstNumber - secondNumber << '\n';
    std::cout << "Product: " << firstNumber * secondNumber << '\n';

        std::cout << "Quotient: " << firstNumber / secondNumber << '\n';
    } else {
        std::cout << "Cannot divide by zero.\n";
    }

    return 0;
}
```

Compile and run:


## Lesson 4: Making decisions with `if`

An `if` statement runs code only when its condition is true. Use `else if` to check another condition, and `else` for the remaining case. Comparisons include `==` (equal), `<` (less than), and `>=` (greater than or equal).

```cpp
%%writefile lesson04.cpp
#include <iostream>

int main() {
    int temperature;

    std::cout << "Enter the temperature: ";
    std::cin >> temperature;

    if (temperature >= 30) {
        std::cout << "It's warm.\n";
    } else if (temperature >= 15) {
        std::cout << "It's mild.\n";
    } else {
        std::cout << "It's cool.\n";
    }

    return 0;
}
```

Compile and run:

```python
!g++ -std=c++17 -Wall -Wextra lesson04.cpp -o lesson04 && ./lesson04
```

## Lesson 5: Repeating work with loops

A `for` loop is handy when you know how many times to repeat something. A `while` loop repeats as long as its condition is true. In the example, the `for` loop adds the numbers from 1 through 5, and the `while` loop counts down.

```cpp
%%writefile lesson05.cpp
#include <iostream>

int main() {
    int total = 0;

    for (int number = 1; number <= 5; number++) {
        total += number;
    }

    std::cout << "Sum from 1 to 5: " << total << '\n';

    int countdown = 3;
    while (countdown > 0) {
        std::cout << countdown << '\n';
        countdown--;
    }

    std::cout << "Go!\n";
    return 0;
}
```

Compile and run:

```python
!g++ -std=c++17 -Wall -Wextra lesson05.cpp -o lesson05 && ./lesson05
```

## Lesson 7: Working with text and lists

A `std::string` stores text. A `std::vector` stores a list of values that can grow or shrink. This example uses a range-based `for` loop to visit each item in a vector. The `const` reference `const std::string&` lets the loop read each word without making a copy.

```cpp
%%writefile lesson07.cpp
#include <iostream>
#include <string>
#include <vector>

int main() {
    std::string greeting = "Hello";
    std::vector<std::string> names = {"Mina", "Leo", "Sam"};

    for (const std::string& name : names) {
        std::cout << greeting << ", " << name << "!\n";
    }

    return 0;
}
```

Compile and run:

```python
!g++ -std=c++17 -Wall -Wextra lesson07.cpp -o lesson07 && ./lesson07
```

## Lesson 8: Grouping data with a class

A class defines a type that can hold related data and functions. Here, `Dog` groups a name with a function, `bark`. `Dog myDog("Pip")` creates a dog and gives it a name.

```cpp
%%writefile lesson08.cpp
#include <iostream>
#include <string>

class Dog {
public:
    std::string name;

    Dog(const std::string& dogName) {
        name = dogName;
    }

    void bark() {
        std::cout << name << " says woof!\n";
    }
};

int main() {
    Dog myDog("Pip");
    myDog.bark();

    return 0;
}
```

Compile and run:

```python
!g++ -std=c++17 -Wall -Wextra lesson08.cpp -o lesson08 && ./lesson08
```

## Lesson 9: Mini-project — a three-question quiz

This small quiz brings together functions, strings, input, and a conditional. The function asks a question, reads the response, and returns `1` for a correct answer or `0` otherwise. Type each answer and press Enter.

```cpp
%%writefile lesson09.cpp
#include <iostream>
#include <string>

int askQuestion(const std::string& question, const std::string& correctAnswer) {
    std::string response;

    std::cout << question << ' ';
    std::getline(std::cin, response);

    if (response == correctAnswer) {
        std::cout << "Correct!\n";
        return 1;
    } else {
        std::cout << "The answer is " << correctAnswer << ".\n";
        return 0;
    }
}

int main() {
    int score = 0;

    score += askQuestion("What language are we learning?", "C++");
    score += askQuestion("How many sides does a triangle have?", "3");
    score += askQuestion("What is 2 + 2?", "4");

    std::cout << "Your score: " << score << " out of 3\n";
    return 0;
}
```

Compile and run:

```python
!g++ -std=c++17 -Wall -Wextra lesson09.cpp -o lesson09 && ./lesson09
```
