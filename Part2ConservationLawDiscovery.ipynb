```python
import numpy as np

def simulate_holographic_universe(potential_func, phi_initial=5.0, steps=1000):
    """
    Simulates a simplified scalar field (inflation) model given a candidate
    potential function V(phi), tracking whether the universe achieves a smooth exit.
    """
    dt = 0.001
    phi = phi_initial
    velocity = 0.0

    scale_factor_history = []
    is_smooth_exit = True

    for _ in range(steps):
        # Numerical approximation of field evolution (Klein-Gordon equation in expanding space)
        # d^2phi/dt^2 + 3H dphi/dt + dV/dphi = 0 (simplified here for demonstration)
        dV_dphi = gradient_potential(potential_func, phi)

        # Simple update rule for the field
        acceleration = -dV_dphi - 0.1 * velocity
        velocity += acceleration * dt
        phi += velocity * dt

        # Track if the field diverges into an uncontrolled runaway (fractal multiverse trap)
        if abs(phi) > 50.0 or np.isnan(phi):
            is_smooth_exit = False
            break

        scale_factor_history.append(phi)

    return is_smooth_exit, scale_factor_history

def gradient_potential(func_name, phi):
    """Numerical derivative of the candidate potential function."""
    eps = 1e-5
    return (evaluate_potential(func_name, phi + eps) - evaluate_potential(func_name, phi - eps)) / (2 * eps)

def evaluate_potential(func_name, phi):
    """Evaluates different mathematical formulations of the inflation potential V(phi)."""
    if func_name == "exponential":
        return np.exp(-0.5 * phi)
    elif func_name == "power_law":
        return phi**4
    elif func_name == "cosmological_toy_bound":
        # A bounded potential designed to suppress eternal inflation
        return phi**2 / (1.0 + 0.1 * phi**2)
    return phi**2

def discover_cosmological_law():
    """
    Searches a library of mathematical potentials to find which one yields
    a globally finite, smooth universe exit condition.
    """
    candidate_potentials = ["exponential", "power_law", "cosmological_toy_bound"]

    print("Evaluating candidate quantum potentials against boundary constraints...\n")

    for candidate in candidate_potentials:
        smooth_exit, history = simulate_holographic_universe(candidate)

        status = "VIABLE: Smooth Exit Discovered!" if smooth_exit else "REJECTED: Leads to Fractal Singularity"
        print(f"[{status}] Potential Model: {candidate}")

# Run the automated cosmological search
discover_cosmological_law()
```
